#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
ncaa_tidydata <- ncaa_data |>
    select(conf_wins_pct, reg_wins, tourney_finish) |>
# Selected the relevant columns for my predictions, keeping only the two predictors and the column to be predicted
    mutate(tourney_finish = factor(tourney_finish,
                                      levels = c("opening_round_loss",
                                            "first_round_loss",
                                            "second_round_loss",
                                            "top_16_loss",
                                            "top_8_loss",
                                            "top_4_loss",
                                            "top_2_loss",
                                            "champ")
                                     ))
# Turned tourney_finish into a factor so R treats it as a category, also reordered the categories (otherwise R orders by which shows up first based on the data).
# Factor was used rather than as_factor since as_factor doesn't allow the convenient reordering.
# No further transformation is needed, as the data is tidy in the form of a single variable per column, a single observation per row, and a single value per cell
ncaa_tidydata_preview <- slice(ncaa_tidydata, 1:6)
# Sliced the first 6 rows of the tidied data
ncaa_tidydata_preview
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
numeric_summary <- ncaa_tidydata |>
    select(conf_wins_pct, reg_wins) |>
    summarize(
        conf_wins_pct_mean = mean(conf_wins_pct, na.rm = TRUE),
        conf_wins_pct_max = max(conf_wins_pct, na.rm = TRUE),
        conf_wins_pct_min = min(conf_wins_pct, na.rm = TRUE),
        reg_wins_mean = mean(reg_wins),
        reg_wins_max = max(reg_wins),
        reg_wins_min = min(reg_wins),
    )

numeric_summary
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table.

```r
na_count <- ncaa_tidydata |>
    summarize(NA_count_conf_wins_pct = sum(is.na(conf_wins_pct)),
              NA_count_reg_wins = sum(is.na(reg_wins)),
              NA_count_tourney_finish = sum(is.na(tourney_finish))
             )

na_count
```
